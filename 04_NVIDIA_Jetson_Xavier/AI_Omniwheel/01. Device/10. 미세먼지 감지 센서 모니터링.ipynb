### 10. 미세먼지 감지 센서 모니터링
#### I2C로 연결된 적외선 온도 측정 센서로 온도를 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C AD/DA Converter 칩 전용 모듈의 내용 중 pcf8591클래스와 AnalogIn 클래스를 추가
import adafruit_pcf8591.pcf8591 as PCF
from adafruit_pcf8591.analog_in import AnalogIn
```

```python
# GPIO핀 사용을 위한 Jetson GPIO 모듈 추가
import Jetson.GPIO as GPIO
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 핀 설정

```python
# 센서 LED가 연결된 핀 정의
Light = 'SPI3_CS1_N'

# 센서 LED가 연결된 핀 출력 설정 및 초기화
GPIO.setup(Light, GPIO.OUT)
GPIO.output(Light, GPIO.HIGH)
```

#### 3) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# PCF8591 칩을 제어하기 위해 클래스 객체 생성
pcf = PCF.PCF8591(i2c)
```

```python
# PCF8591 칩의 0번 핀과 미세먼지 센서가 연결되어 있으므로 0번핀 객체 생성
pcf_in_0 = AnalogIn(pcf, PCF.A0)
```

#### 4) 센서 값 모니터링

```python
while True:
    GPIO.output(Light, GPIO.LOW)

    time.sleep(0.00028)
    raw_value = pcf_in_0.value
    scaled_value = (raw_value / 65535) * pcf_in_0.reference_voltage

    time.sleep(0.00004)
    GPIO.output(Light, GPIO.HIGH)

    time.sleep(0.00968)
    print("미세먼지량 : %.2f"%scaled_value)
```
