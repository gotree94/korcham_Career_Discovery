### 6. Light 센서 모니터링
#### I2C로 연결된 조도 센서로 빛의 세기를 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C 조도 센서 전용 모듈 추가
import adafruit_bh1750
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# Light 센서를 사용하기 위해 객체 생성
light_sensor = adafruit_bh1750.BH1750(i2c)
```

#### 3) 센서 값 모니터링

```python
while True:
    val = light_sensor.lux
    
    print("조도 감지 센서 : %.2f Lux"%val)
    time.sleep(1)
```
