### 1-1. LED 점멸 제어
#### I2C PWM 제어 칩(PCA9685)과 연결된 LED의 ON/OFF를 제어한다. 

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C PWM 제어 칩 전용 모듈 추가
import adafruit_pca9685
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA)
```

```python
# PCA9685 칩을 제어하기 위해 클래스 객체 생성
pca9685 = adafruit_pca9685.PCA9685(i2c)

# PWM 주파수 주기(Hz) 설정
pca9685.frequency = 60
```

#### 3) 동작 제어

```python
# 0~7까지 LED 8개를 순서대로 ON
for index in range(0, 8):
    pca9685.channels[index].duty_cycle = 4095
    time.sleep(0.5)
```

```python
# 7~0까지 LED 8개를 순서대로 OFF
for index in range(8, -1, -1):
    pca9685.channels[index].duty_cycle = 0
    time.sleep(0.5)
```
