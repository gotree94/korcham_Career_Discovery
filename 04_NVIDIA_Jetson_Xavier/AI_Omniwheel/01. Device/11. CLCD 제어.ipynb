### 11. CLCD 제어
#### I2C로 연결된 CLCD에 문자를 출력한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C CLCD 제어 모듈 추가
import libraries.I2C_Charactor_Liquid_Crystal_Display as i2c_lcd
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA)
```

```python
# LCD를 제어하기 위해 클래스 객체 생성
lcd = i2c_lcd.LCD(i2c)
```

#### 3) 동작 제어

```python
# CLCD의 Backlight 1초 간격으로 3회 ON/OFF
for i in range(0, 3):
    lcd.backlight(0)
    time.sleep(1)
    
    lcd.backlight(1)
    time.sleep(1)
```

```python
# CLCD에 문자열 출력
lcd.setCursor(0, 0)
lcd.display_string("Hello World!")

lcd.setCursor(1, 0)
lcd.display_string("Omniwheel")
```

```python
# LCD 화면 초기화
lcd.clear()
```
