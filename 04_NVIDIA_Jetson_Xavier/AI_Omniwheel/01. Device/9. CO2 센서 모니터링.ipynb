### 9. CO2 센서 모니터링
#### I2C로 연결된 CO2 센서로 CO2 가스량을 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C 조도 감지 센서 전용 모듈 추가
import adafruit_ccs811
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# CO2 센서를 사용하기 위해 객체 생성(address - 0x5b)
ccs811 = adafruit_ccs811.CCS811(i2c, 0x5b)
```

#### 3) 센서 값 모니터링

```python
while True:
    eco2 = ccs811.eco2
    tvoc = ccs811.tvoc

    print("CO2: %1.0f PPM" %eco2)
    print("TVOC: %1.0f PPM" %tvoc)

    time.sleep(1)
```
