### 7. 적외선 온도 센서 모니터링
#### I2C로 연결된 적외선 온도 센서로 온도를 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C 적외선 온도 센서 전용 모듈 추가
import adafruit_mlx90614
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# 적외선 온도 센서를 사용하기 위해 객체 생성
mlx = adafruit_mlx90614.MLX90614(i2c)
```

#### 3) 센서 값 모니터링

```python
while True:
    temp_amb = mlx.ambient_temperature
    temp_obj = mlx.object_temperature

    print("Ambent Temp: %.2f" %temp_amb)
    print("Object Temp: %.2f" %temp_obj)
    time.sleep(1)
```
