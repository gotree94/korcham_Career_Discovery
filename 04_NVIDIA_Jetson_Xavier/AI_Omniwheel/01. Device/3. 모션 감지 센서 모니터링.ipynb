### 3. 모션 감지 센서 모니터링
####  GPIO와 연결된 모션감지 센서로 전방의 물체 이동 감지 여부를 측정한다.

#### 1) 모듈 추가

```python
# GPIO핀 사용을 위한 Jetson GPIO 모듈 추가
import Jetson.GPIO as GPIO
GPIO.setwarnings(False)
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 핀 설정

```python
# GPIO 모드 설정
GPIO.setmode(GPIO.TEGRA_SOC)

# 센서가 연결된 핀 정의
Motion = 'SPI3_CS0_N'

# 센서가 연결된 핀 입력 설정
GPIO.setup(Motion, GPIO.IN)
```

#### 3) 센서 값 모니터링

```python
while True:
    value = GPIO.input(Motion)
    
    # 모션 감지가 된 경우
    if(value == 0):
        print("모션 감지")

    time.sleep(0.3)
        
```
