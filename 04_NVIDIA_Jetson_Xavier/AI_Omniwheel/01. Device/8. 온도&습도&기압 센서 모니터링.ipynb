### 8. 온도/습도/기압 센서 모니터링
#### I2C로 연결된 온도/습도/기압 센서로 온도/습도/기압을 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C 온도/습도/기압 감지 센서 전용 모듈 추가
from adafruit_bme280 import basic as adafruit_bme280
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# 온도/습도/기압 센서를 사용하기 위해 객체 생성(address - 0x76)
bme280 = adafruit_bme280.Adafruit_BME280_I2C(i2c, 0x76)
```

#### 3) 센서 값 모니터링

```python
while True:
    temp = bme280.temperature
    humi = bme280.humidity
    press = bme280.pressure

    print("\nTemperature: %0.1f C" %temp)
    print("Humidity: %0.1f %%" %humi)
    print("Pressure: %0.1f hPa" %press)

    time.sleep(1)
```
