### 5. Flame 센서 모니터링
#### I2C ADC/DAC 칩과 연결된 Flame 센서로 불꽃 감지 여부를 측정한다.

#### 1) 모듈 추가

```python
# I2C 통신을 위한 통신 버스 모듈 추가
import busio
import board
```

```python
# I2C AD/DA Converter 칩 전용 모듈의 내용 중 pcf8591클래스와 AnalogIn 클래스를 추가
import adafruit_pcf8591.pcf8591 as PCF
from adafruit_pcf8591.analog_in import AnalogIn
```

```python
# 시간 지연을 위한 time 모듈 추가
import time
```

#### 2) 클래스 객체 생성

```python
# I2C 통신을 이용하기 위해 busio의 I2C 클래스 객체 생성
i2c = busio.I2C(board.SCL, board.SDA, frequency = 100000)
```

```python
# PCF8591 칩을 제어하기 위해 클래스 객체 생성
pcf = PCF.PCF8591(i2c)
```

```python
# PCF8591 칩의 2번 핀과 Flame 센서가 연결되어 있으므로 2번핀 객체 생성
pcf_in_2 = AnalogIn(pcf, PCF.A2)
```

#### 3) 센서 값 모니터링

```python
while True:
    raw_value = pcf_in_2.value

    scaled_value = (raw_value / 65535) * pcf_in_2.reference_voltage
    
    if(scaled_value < 1.6):
        print("Flame 센서 감지됨")

    time.sleep(0.3)
```
