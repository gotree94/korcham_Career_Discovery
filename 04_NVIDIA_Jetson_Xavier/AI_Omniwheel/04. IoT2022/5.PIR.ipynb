```python
import busio
import board
import adafruit_pcf8591.pcf8591 as PCF
from adafruit_pcf8591.analog_in mport Analogln
import time
i2c = busio.I2C(board.SCL_1, board.SDA_1)
pcf = PCF.PCF8591(i2c)
pcf_in_1 = Analogln(pcf, PCF.A1)
while True:
    raw_value = pcf_in_1.value
    scaled_value = (raw_value / 65535) * pcf_in_1.reference_voltage
    if(scaled_value) > 1.6:
        print("PIR 센서 감지됨")
    time.sleep(0.3)
```
