```python
import busio
import board
import adafruit_pcf8591.pcf8591 as PCF
from adafruit_pcf8591.analog_in import Analogln
import time
i2c = busio.I2C(board.SCL_1, board.SDA_1, frequency = 100000)
pcf = PCF.PCF8591(i2c)
pcf_in_2 = Analogln(pcf, PCF.A2)
while True:
    raw_value = pcf_in_2.value
    scaled_value = (raw_value / 65535) * pcf_in_2.reference_voltage
    if(scaled_value < 1.6):
        print("Flame 센서 감지됨")
        
    time.sleep(0.3)
```
