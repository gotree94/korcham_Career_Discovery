```python
import busio
import board
import adafruit_mlx90614
import time

i2c=busio.I2C(board.SCL_1,board.SDA_1,frequency=100000)
mlx = adafruit_mlx90614.MLX90614(i2c)
while True:
    temp_amb=mlx.ambient_temperature
    temp_obj=mlx.object_temperature
    print("Ambent Temp: %2.f" %temp_amb)
    print("Object Temp: %2.f" %temp_obj)
    time.sleep(1)
```
