```python
import time
import busio
import board
import adafruit_ccs811
i2c = busio.I2C(board.SCL_1, board.SDA_1, frequency = 100000)
ccs811 = adafruit_ccs811.CCS811(i2c, 0x5b)
while True:
    eco2 = ccs811.eco2
    tvoc = ccs811.tvoc
    print("CO2: %1.0f PPM" %eco2)
    print("TVOC: %1.0f PPM" %tvoc)
    time.sleep(1)
```
