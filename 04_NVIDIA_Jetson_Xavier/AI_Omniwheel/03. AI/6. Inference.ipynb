```python
import os
import cv2
import time

import bready.camera_utils.PyCamera as PyCamera
from bready.object_detection_tools import ObjectDetector

import ipywidgets.widgets as widgets
import IPython.display
```

```python
camera = PyCamera.URLCamera(
    camera_url = "tcp://127.0.0.1:5000",
    live_thread = True
)
```

```python
obj_model = ObjectDetector.Tester(
    'optimization',
    'detection_classes.txt',
    ["image_tensor"]
)
```

```python
image_widget = widgets.Image(format='jpeg')
IPython.display.display(image_widget)

while(True):
    try:
        image = camera.get_frame()

        if(image is None):
            break

        obj_model.execute(image)
        
        if obj_model.getDetectedCount('person') > 0:
            print("사람이 감지되었습니다")
            
        result_image = obj_model.getResultImage()
        image_widget.value = cv2.imencode(".jpeg", result_image)[1].tostring()
        time.sleep(0.01)
        
    except KeyboardInterrupt:
        print("종료")
        break
```
