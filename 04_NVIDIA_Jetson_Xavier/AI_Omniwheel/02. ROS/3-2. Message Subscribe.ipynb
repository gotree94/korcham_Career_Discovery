# 3-2. ROS Message Subscribe

#### 모듈 추가

```python
import rospy
from omniwheel_project.msg import Numb
```

#### 데이터를 수신 처리

```python
def callback(data):
    rospy.loginfo(rospy.get_caller_id() + "I heard %d", data.number)
```

#### 노드 생성

```python
rospy.init_node('listener', anonymous=True)
```

#### Subscriber 객체 생성

```python
rospy.Subscriber("msgChatter", Numb, callback)
```
