# 3-1. ROS Message Publish

#### 모듈 추가

```python
import rospy
from omniwheel_project.msg import Numb
```

#### 노드 생성

```python
rospy.init_node('talker', anonymous=True)
```

#### Publisher 객체 생성

```python
pub = rospy.Publisher('msgChatter', Numb, queue_size=10)
```

#### 데이터 발행 주기 지정

```python
rate = rospy.Rate(10)
```

#### 변수 선언 및 초기화

```python
num = 0
```

#### 문자열 전송

```python
while not rospy.is_shutdown():
    rospy.loginfo(num)
    
    pub.publish(num)
    
    rate.sleep()
    
    num += 1
```
