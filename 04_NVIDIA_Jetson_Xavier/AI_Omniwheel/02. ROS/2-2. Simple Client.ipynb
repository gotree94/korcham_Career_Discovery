# 2-2. ROS SIMPLE Client

#### 모듈 추가

```python
import rospy
from omniwheel_project.srv import AddTwoInts,AddTwoIntsResponse
```

#### 서비스 실행 대기

```python
rospy.wait_for_service('add_two_ints')
```

#### 서비스를 요청할 handle 선언

```python
add_two_ints = rospy.ServiceProxy('add_two_ints', AddTwoInts)
```

#### 서비스 요청 및 응답 데이터 확인

```python
result = add_two_ints(2, 3)
print("Service Answer : %s"%result)
```
