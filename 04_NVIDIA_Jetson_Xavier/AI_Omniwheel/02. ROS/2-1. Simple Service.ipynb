# 2-1. ROS SIMPLE Service

#### 모듈 추가

```python
import rospy
from omniwheel_project.srv import AddTwoInts,AddTwoIntsResponse
```

#### 서비스 요청 수신시 처리할 함수 선언

```python
def handle_add_two_ints(req):
    print("Returning [%s + %s = %s]"%(req.a, req.b, (req.a + req.b)))
    
    return AddTwoIntsResponse(req.a + req.b)
```

#### 노드 생성

```python
rospy.init_node('add_two_ints_server', anonymous=True)
```

#### 요청 받을 Server 생성

```python
server = rospy.Service('add_two_ints', AddTwoInts, handle_add_two_ints)
```
